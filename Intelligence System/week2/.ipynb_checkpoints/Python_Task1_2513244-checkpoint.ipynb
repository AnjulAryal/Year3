## Question A6: Function: palindrome checker  `[5 marks]`

Write a function `is_palindrome(text)` that returns `True` if the text reads the same forwards and backwards, and `False` otherwise.

The check must **ignore upper/lower case, spaces and punctuation**. Only letters and digits count.

```
is_palindrome("Never odd or even")             -> True
is_palindrome("A man, a plan, a canal: Panama") -> True
is_palindrome("Python")                         -> False
```

**Hint:** build a cleaned string first (`isalnum()`, `lower()`), then compare it with its reverse.

````python
# ✏️ Write your answer below

def is_palindrome(text):
    cleaned = ""

    for char in text:
        if char.isalnum():
            cleaned += char.lower()

    return cleaned == cleaned[::-1]
   

print(is_palindrome("Never odd or even"))
print(is_palindrome("A man, a plan, a canal: Panama"))
print(is_palindrome("Python") )
````

````python
# Run this cell to test your answer
try:
    check("'Never odd or even'", is_palindrome("Never odd or even") is True)
    check("'A man, a plan, a canal: Panama'", is_palindrome("A man, a plan, a canal: Panama") is True, "ignore punctuation")
    check("'Was it a car or a cat I saw?'", is_palindrome("Was it a car or a cat I saw?") is True)
    check("'12321'", is_palindrome("12321") is True, "digits count too")
    check("'Python' is not a palindrome", is_palindrome("Python") is False)
    check("'Herald College' is not a palindrome", is_palindrome("Herald College") is False)
except NameError as e:
    print('⚠️ Something is not defined yet:', e, '(run your answer cell first)')
except Exception as e:
    print('❌ Your code raised an error while testing:', type(e).__name__, '-', e)
````

## Question A7: Function: Caesar cipher  `[6 marks]`

A **Caesar cipher** hides a message by shifting each letter along the alphabet. With a shift of 3, `a` becomes `d`, `b` becomes `e`, and `z` wraps around to `c`.

Write a function `caesar_shift(text, shift)` that returns the shifted text. Rules:

* Upper-case letters stay upper-case and lower-case letters stay lower-case.
* Anything that is not a letter (spaces, digits, punctuation) is left unchanged.
* The alphabet wraps around (`xyz` shifted by 3 gives `abc`).
* A negative shift moves backwards, so shifting by `-5` undoes a shift of `5`.

```
caesar_shift("abc", 1)            -> "bcd"
caesar_shift("Hello, World!", 5)  -> "Mjqqt, Btwqi!"
```

**Hint:** `ord()` gives the number of a character, `chr()` turns a number back into a character, and `%` helps with wrapping.

````python
# ✏️ Write your answer below

def caesar_shift(text, shift):
    result = ""

    for char in text:
        if char.isupper():
            shifted = (ord(char) - ord('A') + shift) % 26
            result += chr(shifted + ord('A'))

        elif char.islower():
            shifted = (ord(char) - ord('a') + shift) % 26
            result += chr(shifted + ord('a'))

        else:
            result += char

    return result

caesar_shift("abc", 1) 
caesar_shift("Hello, World!", 5) 
````

````python
# Run this cell to test your answer
try:
    check("simple shift", caesar_shift("abc", 1) == "bcd")
    check("wrap around", caesar_shift("xyz", 3) == "abc", "use % 26")
    check("case and punctuation", caesar_shift("Hello, World!", 5) == "Mjqqt, Btwqi!")
    check("negative shift", caesar_shift("Mjqqt, Btwqi!", -5) == "Hello, World!")
    check("large shift (31 is the same as 5)", caesar_shift("Hello, World!", 31) == "Mjqqt, Btwqi!")
    check("digits unchanged", caesar_shift("Room 101", 2) == "Tqqo 101")
except NameError as e:
    print('⚠️ Something is not defined yet:', e, '(run your answer cell first)')
except Exception as e:
    print('❌ Your code raised an error while testing:', type(e).__name__, '-', e)
````
