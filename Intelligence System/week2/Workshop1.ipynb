```python
from coppeliasim_zmqremoteapi_client import RemoteAPIClient
```

```python
import os
import time


desiredPath = "/Users/anjularyal/Desktop/Year 3/Intelligence System/week2"
if not os.path.isdir(desiredPath):
    raise FileNotFoundError(f"Create the workshop folder first: {desiredPath}")
os.chdir(desiredPath)


print("Current working directory:", os.getcwd())


print("Program started")


try:
    client = RemoteAPIClient("127.0.0.1", 23000)
    sim = client.require("sim")
    print("Connected to CoppeliaSim")


except Exception as error:
    print("Connection failed!!")
    print(error)
    raise
```

```python
print("Program started")


try:
    objs = sim.getObjectsInTree(
        sim.handle_scene, # whole scene
        sim.handle_all, # all object types
        0
    )


    print("Number of objects in the scene:", len(objs))


except Exception as error:
    print("Could not retrieve scene objects:")
    print(error)


time.sleep(2)
```

```python
def find_component(alias):
    handle = sim.getObject("/" + alias, {"noError": True}) # coppeliasim has given objects an id i.e. handle
    if handle != -1: # if the object was found, use its ID
        return handle
    objects = sim.getObjectsInTree(sim.handle_scene, sim.handle_all, 0)
    matches = [h for h in objects if sim.getObjectAlias(h) == alias]
    if len(matches) != 1:
        raise RuntimeError(
            f"Expected one {alias}, found {len(matches)}. Check the scene hierarchy."
        )
    return matches[0]




def attach_components():
    names = ["DynamicLeftJoint", "DynamicRightJoint",
             "LeftSensor", "MiddleSensor", "RightSensor"]
    return tuple(find_component(name) for name in names) #This line calls find_component() once for each name in names and puts the returned handles into a tuple in the same order.




def stop_motors(left, right):
    try:
        sim.setJointTargetVelocity(left, 0.0)
    finally:
        sim.setJointTargetVelocity(right, 0.0)




def require_running():
    if sim.getSimulationState() in (sim.simulation_stopped, sim.simulation_paused):
        raise RuntimeError("Press Play in CoppeliaSim before running this task.")




require_running()
leftMotor, rightMotor, leftSensor, midSensor, rightSensor = attach_components()
stop_motors(leftMotor, rightMotor)
print("Motor handles:", leftMotor, rightMotor)
print("Sensor handles:", leftSensor, midSensor, rightSensor)
time.sleep(2)
print("...done")
```

```python
print("Program started")
require_running()
leftMotor, rightMotor, leftSensor, midSensor, rightSensor = attach_components()


try:
    sim.setJointTargetVelocity(leftMotor, 0.6)
    sim.setJointTargetVelocity(rightMotor, -0.6)


    time.sleep(6)
finally:
    stop_motors(leftMotor, rightMotor)
    time.sleep(0.5)


print("...done")
```

```python
print("Program started")
require_running()
leftMotor, rightMotor, leftSensor, midSensor, rightSensor = attach_components()


try:
    sim.setJointTargetVelocity(rightMotor, 0.0)
    sim.setJointTargetVelocity(leftMotor, 0.6)
    time.sleep(5)


    # Move forward:
    startTime = time.time()


    sim.setJointTargetVelocity(leftMotor, 0.6)
    sim.setJointTargetVelocity(rightMotor, 0.6)


    while time.time() - startTime < 15:
        # Read/process sensor data here:
        time.sleep(0.005)
finally:
    stop_motors(leftMotor, rightMotor)
    time.sleep(1)


print("...done")
```

```python
print("Program started")
require_running()
leftMotor, rightMotor, leftSensor, midSensor, rightSensor = attach_components()


try:
    sim.setJointTargetVelocity(rightMotor, 0.0)
    sim.setJointTargetVelocity(leftMotor, 0.6)
    time.sleep(5)


    sensorReading = [None, None, None]
    startTime = time.time()


    # Move forward:
    sim.setJointTargetVelocity(leftMotor, 0.6)
    sim.setJointTargetVelocity(rightMotor, 0.6)


    while time.time() - startTime < 16:
        # readVisionSensor returns detection state first.
        leftResult = sim.readVisionSensor(leftSensor)
        midResult = sim.readVisionSensor(midSensor)
        rightResult = sim.readVisionSensor(rightSensor)


        sensorReading[0] = leftResult[0] if isinstance(leftResult, (tuple, list)) else leftResult
        sensorReading[1] = midResult[0] if isinstance(midResult, (tuple, list)) else midResult
        sensorReading[2] = rightResult[0] if isinstance(rightResult, (tuple, list)) else rightResult


        print(sensorReading)
        time.sleep(0.5)
finally:
    stop_motors(leftMotor, rightMotor)
    time.sleep(1)


print("...done")
```
