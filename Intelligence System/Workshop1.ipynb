**Goal:** Control a cube in CoppeliaSim from Python, move it using loops, and print log messages inside CoppeliaSim.

## Step 1: Connect Python to CoppeliaSim
This creates a connection between the notebook and CoppeliaSim so Python can send commands to it.

```python
# Import the class that creates the connection to CoppeliaSim
from coppeliasim_zmqremoteapi_client import RemoteAPIClient

# Create a client object; this opens the connection to the running CoppeliaSim
client = RemoteAPIClient()

# Get the 'sim' API, which gives us all the simulation functions
sim = client.require('sim')

# Confirm the connection worked
print("Connected!")

# Print the current simulation time (0.0 because the simulation hasn't started)
print("Simulation time:", sim.getSimulationTime())
```

## Step 2: Get the Cube
This finds the cube in the scene and stores its **handle** (an ID number). Python uses the handle to control the cube.

```python
# Search the scene for the object named 'Cube'
cube = sim.getObject('/Cube')

# Print the handle; if you see a number, the cube was found
print("Cube handle:", cube)
```

## Step 3: Read the Cube's current position
The position is a list [X, Y, Z] in metres. The value -1 means "relative to the world".

```python
# Get the cube's position relative to the world (-1 = world frame)
position = sim.getObjectPosition(cube, -1)

# Show the result as [X, Y, Z]
print("Cube position:", position)
```

## Step 4: Move the Cube
This moves the cube directly to X = 1, Y = 0, Z = 0.5.

```python
# Set position: object handle, reference frame (-1 = world), [X, Y, Z]
sim.setObjectPosition(cube, -1, [1, 0, 0.5])

# Confirmation message
print("Cube moved!")
```

## Step 5: Move gradually (like a robot)
A `for` loop moves the cube in small steps so the motion looks smooth.

### 5a. Move along the X-axis

```python
# Import time so we can pause between steps
import time

# Repeat 11 times: x = 0, 1, 2, ... 10
for x in range(0, 11):
    # X goes 0.0, 0.1, ... 1.0; Y stays 0; Z stays 0.5
    sim.setObjectPosition(cube, -1, [x * 0.1, 0, 0.5])

    # Wait 0.2 seconds so the motion is visible
    time.sleep(0.2)
```

### 5b. Move along the Y-axis

```python
# Repeat 11 times: y = 0, 1, ... 10
for y in range(0, 11):
    # X stays 1; Y goes 0.0 to 1.0; Z stays 0.5
    sim.setObjectPosition(cube, -1, [1, y * 0.1, 0.5])

    # Short pause between steps
    time.sleep(0.2)
```

### 5c. Move along the Z-axis (upwards)

```python
# Repeat 11 times: z = 0, 1, ... 10
for z in range(0, 11):
    # X = 1 and Y = 1 stay fixed; Z rises from 0.5 to 1.5
    sim.setObjectPosition(cube, -1, [1, 1, 0.5 + z * 0.1])

    # Short pause between steps
    time.sleep(0.2)
```

## Step 6: first exercise (all in one cell)
Connect, read the position, move, wait, then read the position again.


```python
# Imports
from coppeliasim_zmqremoteapi_client import RemoteAPIClient
import time

# Connect to CoppeliaSim
client = RemoteAPIClient()
sim = client.require('sim')
print("Connected to CoppeliaSim")

# Get the cube handle
cube = sim.getObject('/Cube')

# Read and print the starting position
position = sim.getObjectPosition(cube, -1)
print("Initial position:", position)

# Move the cube to X=1, Y=0, Z=0.5
sim.setObjectPosition(cube, -1, [1, 0, 0.5])

# Wait 2 seconds
time.sleep(2)

# Read and print the new position
position = sim.getObjectPosition(cube, -1)
print("New position:", position)
```




**Task:** Move the cube to different positions using a loop and show a log message inside CoppeliaSim.

**Tips used:**
- a. Store positions in a list.
- b. Use a for loop to go through them.
- c. Use `sim.addLog(sim.verbosity_scriptinfos, "message")` to log inside CoppeliaSim.

The cube moves in a square path.


```python
import time

# Tip (a): list of positions; each item is one [X, Y, Z] point
positions = [
    [0.5, 0.0, 0.5],   # Point 1
    [0.5, 0.5, 0.5],   # Point 2
    [0.0, 0.5, 0.5],   # Point 3
    [0.0, 0.0, 0.5],   # Point 4 (back to start)
]

# Tip (b): loop through the list; i = index (0,1,2,3), pos = the position
for i, pos in enumerate(positions):

    # Move the cube to this position
    sim.setObjectPosition(cube, -1, pos)

    # Tip (c): write a log message inside CoppeliaSim (bottom console)
    sim.addLog(sim.verbosity_scriptinfos, f"Step {i+1}: Cube moved to {pos}")

    # Also print in Jupyter so you can see it here
    print(f"Step {i+1}: Cube moved to {pos}")

    # Wait 1 second before the next move
    time.sleep(1)

# Final log message after the loop ends
sim.addLog(sim.verbosity_scriptinfos, "Movement finished!")
print("Movement finished!")
```
